```python
import os
import warnings

import pandas as pd
```

```python
with warnings.catch_warnings():
    warnings.filterwarnings("ignore", category=pd.errors.DtypeWarning)
    df_combined = pd.read_csv('../output/combined_part_7.csv')

df_combined['date'] = pd.to_datetime(df_combined['date'])
df_countries = df_combined.query("aggregation_level == 0").copy()
```

##### Liczba nowych zachorowań:

```python
cases_columns = [
    'date', 'country_name',
    'new_confirmed', 'cumulative_confirmed'
]

df_countries['new_confirmed'] = df_countries['new_confirmed'].astype('Int64')
df_countries['cumulative_confirmed'] = df_countries['cumulative_confirmed'].astype('Int64')

df_cases = df_countries[cases_columns].copy()

os.makedirs('covid_cases_sum_by_country/input', exist_ok=True)
df_cases.to_csv('covid_cases_sum_by_country/input/cases.csv', index=False)
```

Liczba nowych śmierci:

```python
deaths_columns = [
    'date', 'country_name',
    'new_deceased', 'cumulative_deceased'
]

df_countries['new_deceased'] = df_countries['new_deceased'].astype('Int64')
df_countries['cumulative_deceased'] = df_countries['cumulative_deceased'].astype('Int64')

df_deaths = df_countries[deaths_columns].copy()

os.makedirs('covid_deaths_max_by_country/input', exist_ok=True)
df_deaths.to_csv('covid_deaths_max_by_country/input/deaths.csv', index=False)
```

Liczba nowych szczepień:

```python
vaccinations_columns = [
    'date', 'country_name',
    'new_persons_vaccinated', 'cumulative_persons_vaccinated'
]

df_countries['new_persons_vaccinated'] = df_countries['new_persons_vaccinated'].astype('Int64')
df_countries['cumulative_persons_vaccinated'] = df_countries['cumulative_persons_vaccinated'].astype('Int64')

df_vaccinations = df_countries[vaccinations_columns].copy()

os.makedirs('total_covid_vaccination_days_by_country/input', exist_ok=True)
df_vaccinations.to_csv('total_covid_vaccination_days_by_country/input/vaccinations.csv', index=False)
```

Trendy wyszukiwań:

```python
trends_columns = [
    'date', 'country_name',
    'search_trends_alcoholism', 'search_trends_anxiety',
    'search_trends_depression', 'search_trends_insomnia'
]

df_trends = df_countries[trends_columns].copy()

os.makedirs('search_trends_mean_std_by_country/input', exist_ok=True)
df_trends.to_csv('search_trends_mean_std_by_country/input/trends.csv', index=False)
```
